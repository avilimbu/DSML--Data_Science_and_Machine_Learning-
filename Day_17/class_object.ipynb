```python
class Character:
  def __init__(self, health, damage, speed): #instace method (constructor)
    self.health = health #instance attributes
    self.damage = damage
    self.speed = speed

  def dobule_speed(self): #instance method
    self.speed *=2
    return self.speed
```

```python
obj1 = Character(100,50,40)
print(obj1.speed)
print(obj1.dobule_speed())

obj2 = Character(30,40,50)
obj2.dobule_speed()
```

```python
class Product:
  tax_rate = 0.13 #class attributes

  def __init__(self, name, price):
    self.name = name
    self.price = price

p1 = Product("Laptop", 50000)
p1.tax_rate
```

```python
p2 = Product("Phone", 200000)
p2.tax_rate
```

```python
#static method
class Calculator:
  #A uitlity function that logically belong to the class but doenot need self or cls
  @staticmethod
  def add(a,b):
    return a+b

Calculator.add(3,4)
```

```python
class EmailValidator:
  @staticmethod
  def is_valid(email):
    return "@" in email

EmailValidator.is_valid("test@gmail.com")
```

```python
#class method
#works with the class, not an invidual object
class Student:
  school = "ABC" #class attributes
  @classmethod  #class metthod
  def get_school(cls):
    print(cls.school)

Student.get_school()
```
