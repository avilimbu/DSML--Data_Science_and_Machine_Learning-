```python
# magic number
class BankCustomer:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def __str__(self):
        return f"{self.name} is {self.age} years old"
    # called by repr
    def __repr__(self):
        return f"BankCustomer('{self.name}',{self.age})"
    #called by len
    def __len__(self):
        return self.age
    #makes object callable like a function
    def __call__(self, gender ="Male"):
        return f"{self.name} is {self.age} years old {gender}"
customer = BankCustomer("roktavor",33)
```

```python
#str
print(customer)
```

```python
print(repr(customer))
```

```python
#len
print(len(customer))
```

```python
#call
print(customer())
```
