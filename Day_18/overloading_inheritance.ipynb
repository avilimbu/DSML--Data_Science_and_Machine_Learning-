```python
class Polygon:
  def __init__(self,name, sides):
    self.name = name
    self.sides = sides

  def display(self):
    print(f"{self.name} has {self.sides} sides.")

triangle = Polygon("Triangle", 3)
square = Polygon("Square", 4)
pentagon = Polygon("Pentagon", 5)
hexagon = Polygon("Hexagon", 6)

# Displaying information
triangle.display()
square.display()
pentagon.display()
hexagon.display()
```

```python
#method overloading | Using default argument as traditional method overloading is not supported by python
class Calculator:
  def add(self,a,b,c=0):
    return a+b+c

first = Calculator()
first.add(7,8)
```

```python
second = Calculator()
second.add(4,5,9)
```

```python
#method overloadinng: using positional argument
class Demo:
    def add(self, *numbers):
        return sum(numbers)

obj = Demo()

print(obj.add(5))
print(obj.add(5, 10))
print(obj.add(5, 10, 20))
print(obj.add(1, 2, 3, 4, 5))
```

```python
#single inheritance
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

d = Dog()
d.sound()
d.bark()
```

```python
#multiple inheritance
class Father:
    def skills1(self):
        print("Driving")

class Mother:
    def skills2(self):
        print("Cooking")

class Child(Father, Mother):
    pass

c = Child()
c.skills1()
c.skills2()
```

```python
#multilevel inheritance
class Grandparent:
    def show1(self):
        print("Grandparent")

class Parent(Grandparent):
    def show2(self):
        print("Parent")

class Child(Parent):
    def show3(self):
        print("Child")

c = Child()
c.show1()
c.show2()
c.show3()
```

```python
#hierarchical inheritance
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

class Cat(Animal):
    def meow(self):
        print("Cat meows")

d = Dog()
c = Cat()

d.eat()
d.bark()

c.eat()
c.meow()
```

```python
# Hybrid inheritance
class Animal:
  def live(self):
    print("Falls under Animal")

class Carnivour(Animal):
  def eats_meat(self):
    print("Eats meat to survive")

class Hervibour(Animal):
  def eats_grass(self):
    print("Eats grass to survive")

class Omnivour(Carnivour, Hervibour):
  pass

class Bear(Omnivour):
  def name(self):
    print("Bear")

bear = Bear()
bear.name()
bear.eats_grass()
bear.eats_meat()
bear.live()
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show(self):
      print(f"Name: {self.name}, Age: {self.age}")


class Teacher(Person):
    def __init__(self, name, age, subject):
        Person.__init__(self, name, age)
        self.subject = subject

    def show(self):
        print(f"Name: {self.name}, Age: {self.age}, Subject: {self.subject}")


class Student(Person):
    def __init__(self, name, age, student_id):
        Person.__init__(self, name, age)
        self.student_id = student_id

    def show(self):
        print(f"Name: {self.name}, Age: {self.age}, Student ID: {self.student_id}")


class TeachingAssistance(Teacher, Student):
    def __init__(self, name, age, subject, student_id, department, salary):
        Teacher.__init__(self, name, age, subject)
        Student.__init__(self, name, age, student_id)
        self.department = department
        self.salary = salary

    def show(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Subject: {self.subject}")
        print(f"Student ID: {self.student_id}")
        print(f"Department: {self.department}")
        print(f"Salary: {self.salary}")


TA = TeachingAssistance(
    "Tony", 22, "Python", 101, "dsml", 30000
)

TA.show()
```
