```python
import numpy as np
```

```python
matrix_1 = np.array([
    [1,2],
    [3,4]
])
matrix_2 = np.array([
    [5,6],
    [2,3]
])

#matrix addition | element wise
result = matrix_1 + matrix_2
print(result)
```

```python
#matrix subtraction | element wise
result = matrix_1 - matrix_2
print(result)
```

```python
# matrix multiplication
result = np.dot(matrix_1, matrix_2)
print(result)
```

```python
result = np.matmul(matrix_1, matrix_2)
print(result)
```

```python
# element wise operation with scalar
scalar = 3
print(matrix_1)
print("-----")
reult_add = matrix_1 + scalar
print(reult_add)
```

```python
print(matrix_1)
print("-----")
reult_mul = matrix_1 * scalar
print(reult_mul)
```

**Broadcasting in NumPy**

Broadcasting in NumPy allows operations to be performed on matrices of different shapes, as long as the shapes are compatible. When performing an element-wise operation between matrices of different shapes, NumPy automatically "broadcasts" the smaller matrix across the larger one.

For example, a 1D array can be added to a 2D matrix, and NumPy will repeat the 1D array across all rows of the matrix to perform the operation. However, there are specific rules that determine whether broadcasting is possible.

Instead of requiring arrays to have exactly the same dimensions, NumPy automatically **expands the smaller array** to match the larger one when possible.

```python
# broadcasting
matrix_3 = np.array([
    [1,2],
    [3,4]
])

array_1 = np.array([3,4])

result = matrix_3 + array_1
print(result)
```

```python
# computing the determinant
matrix_4 = np.array([
    [1,2],
    [3,4]
])

det = np.linalg.det(matrix_4)
print(det)
```

```python
# inverse
matrix_5 = np.array([
    [1,2],
    [5,4]
])

inv = np.linalg.inv(matrix_5)
print(inv)
```

```python
# Linear Algebra ||
# AX = B
# 2x + y = 5
# x + 3y = 6

a = np.array([
    [2,1],
    [1,3]
])

b = np.array([5,6])
x = np.linalg.solve(a,b)
print(x)
```

```python
# x+y+z =6
# 2x + 5y + z = 4
# 2x + 3y + 8z = 5

a = np.array([
    [1,1,1],
    [2,5,1],
    [2,3,8]
])

b = np.array([6,4,5])
x = np.linalg.solve(a,b)
print(x)
```
