```python
# central Tendency
from scipy import stats
import matplotlib.pyplot as plt
```

```python
x = np.random.standard_t(500,size=1000)
x[:15]
```

```python
# mean
_mean = np.mean(x)
print(f"Mean of x: {_mean}")
```

```python
plt.hist(x, bins = 20,color = "blue")
```

```python
plt.axvline(x =_mean, color = "red")
```

```python
plt.hist(x, bins = 20,color = "blue")
plt.axvline(x =_mean, color = "red")
```

```python
# median
_median = np.median(x)
print(f"Median of x: {_median}")
```

```python
plt.hist(x, bins = 20,color = "blue")
plt.axvline(x =_median, color = "red")
```

```python
_mode = stats.mode(x).mode
print(f"Mode of x: {_mode}")
```

```python
plt.hist(x, bins = 20,color = "blue")
plt.axvline(x =_mode, color = "red")
plt.show()
```
