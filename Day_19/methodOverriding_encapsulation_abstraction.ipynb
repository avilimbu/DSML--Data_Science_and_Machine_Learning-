```python
#method overriding
class Parent:
  def MYmethod(self):
    print("Calling from parent")
class Child(Parent):
  def MYmethod(self):
    super().MYmethod()
    print("calling from child")

c = Child()
c.MYmethod()
```

```python
class Employee:
  def __init__(self,name, salary):
    self.name = name
    self.salary = salary

  def getName(self):
    return self.name
  def getSalary(self):
    return self.salary

class SalesOffice(Employee):
  def __init__(self, name, salary,inc):
    super().__init__(name, salary)
    self.incr = inc

  def getSalary(self):
    return self.salary + self.incr
e1 = Employee("Tony Stark", 30000)
print(f"Salary of {e1.getName()} is RS. {e1.getSalary()}")
```

```python
s1 = SalesOffice("Chopper",30000, 1500)
print(f"Salary of {s1.getName()} is RS. {s1.getSalary()}")
```

```python
class Employee:
    def __init__(self, name, salary, password):
        self.name = name          # Public
        self._salary = salary     # Protected
        self.__password = password  # Private

    # Public method
    def display(self):
        print(f"Name: {self.name}")
        print(f"Salary: {self._salary}")
        print(f"Password: {self.__password}")


class Manager(Employee):
    def show_details(self):
        print(f"Name: {self.name}")
        print(f"Salary: {self._salary}")

        print(f"Password: {self._Employee__password}")


emp = Employee("Tony Stark", 50000, "ironman123")

print("Accessing from outside the class:")
print(emp.name)
print(emp._salary)

print(emp._Employee__password)

print("\nCalling display()")
emp.display()

print("\nAccessing from subclass:")
mgr = Manager("Steve Rogers", 60000, "captain123")
mgr.show_details()
```

```python
#class task
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: Rs. {amount}")
        else:
            print("Invalid deposit amount.")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount.")
        elif amount > self.__balance:
            print("Insufficient balance.")
        else:
            self.__balance -= amount
            print(f"Withdrawn: Rs. {amount}")

    def checkBalance(self):
        return self.__balance


class SavingAccount(BankAccount):
    def __init__(self, name, balance, interest):
        super().__init__(name, balance)
        self.interest = interest

    def calculateInterest(self):
        balance = self.checkBalance()
        interest_amount = (self.interest / 100) * balance
        return balance + interest_amount


s1 = SavingAccount("rodri", 15000, 6)

print("Current Balance:", s1.checkBalance())

s1.deposit(2000)
print("Balance:", s1.checkBalance())

s1.withdraw(3000)
print("Balance:", s1.checkBalance())


print("Balance after Interest:", s1.calculateInterest())
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
  @abstractmethod
  def make_sound(self):
    pass

class Dog(Animal):
  def make_sound(self):
    print("Woof!")

class Cat(Animal):
  def make_sound(self):
    print("Meow!")
```

```python
dog =Dog()
dog.make_sound()
```

```python
cat = Cat()
cat.make_sound()
```
