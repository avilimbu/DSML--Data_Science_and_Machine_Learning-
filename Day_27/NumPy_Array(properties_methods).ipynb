```python
# sorting
unsorted = np.array([2,3,1,0,8,7,9])
print(np.sort(unsorted))
```

```python
# sorting in 2D
two_d_unsorted = np.array([
    [1, 4, 2],
    [4, 2, 6],
    [0,4,1]
])

print(np.sort(two_d_unsorted))
```

```python
print(np.sort(two_d_unsorted, axis=0)) # row
```

```python
print(np.sort(two_d_unsorted, axis=1)) # column
```

```python
# filter
numbers = np.array([1,2,3,4,5,6,7,8])
even_num = numbers[numbers % 2 == 0]
print(even_num)
```

```python
#filter with mask
mask = numbers > 5
print(numbers[mask])
```

```python
numbers[numbers>5] = 1000
print(numbers)
```

```python
# np.where
where_result = np.where(numbers % 2 == 0, "even", "odd")
print(where_result)
```

```python
where_res = np.where(numbers ==5)
print(where_res)
```

```python
num = np.array([1,2,6,5,34,65])
conditional_array = np.where(num<6, num*2,num)
print(conditional_array)
```

```python
a = np.array([
    [1,2,3],
    [4,5,6]
])
b = np.array([11,2,5])
add_row = np.vstack((a,b))
print(add_row)
```

```python
a = np.array([
    [1,2,3],
    [4,5,6]
])
b = np.array([
    [11],
    [12]
])
add_column = np.hstack((a,b))
print(add_column)
```
