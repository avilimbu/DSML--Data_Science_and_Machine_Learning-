Building a survival profile

```python
# analyzing survival by zender
gender_survival = titanic_df.groupby("Sex")["Survived"].mean() * 100

print("Survival by Gender:")
print(gender_survival)
```

```python
# survival by passenger class
class_survival = titanic_df.groupby("Pclass")["Survived"].mean() * 100

print("\nSurvival by Passenger Class:")
print(class_survival)
```

```python
# creating the age grpup
titanic_df["AgeGroup"] = pd.cut(
    titanic_df["Age"],
    bins=[0, 12, 18, 35, 60, 100],
    labels=["Child", "Teenager", "Young Adult", "Adult", "Senior"]
)
```

```python
# sruvival by age group
age_survival = titanic_df.groupby("AgeGroup", observed=True)["Survived"].mean() * 100

print("\nSurvival by Age Group:")
print(age_survival)
```

```python
# creating the fare group with bins
titanic_df["FareGroup"] = pd.cut(
    titanic_df["Fare"],
    bins=[0, 10, 20, 30, 100],
    labels=["Low", "Medium", "High", "Very High"]
)

print(titanic_df[["Fare", "FareGroup"]].head())
fare_sruvival = titanic_df.groupby("FareGroup", observed = True)["Survived"].mean() * 100

print("\nSurvival by Fare Group:")
print(fare_sruvival)
```

```python
# finding the strongest commbined profile
profile = (
    titanic_df.groupby(["Sex", "Pclass", "AgeGroup", "FareGroup"], observed = True)
    .agg(
        sruvival_rate = ("Survived", "mean"),
        Passenger_count = ("Survived", "count")
    )
    .sort_values(by = "sruvival_rate", ascending = False)
    )

profile["survival_rate"] = profile["sruvival_rate"] * 100
profile.head(10)
```

```python
reliable_profiles = profile[
    profile["Passenger_count"] >= 10
]

print(reliable_profiles.head(10))
```

Embarkation prot analysis

```python
# counting passenger from each prot
port_counts = titanic_df["Embarked"].value_counts()

print("Passenger Count by Port of Embarkation:")
print(port_counts)
```

```python
# calculating the survival rate for each prot
port_sruvival = titanic_df.groupby("Embarked")["Survived"].mean() * 100

print("\nSurvival Rate by Port of Embarkation:")
print(port_sruvival)
```

```python
# Finding the port with highest survival rate
highest_survival_port = port_sruvival.idxmax()
highest_survival_rate = port_sruvival.max()

print("\nPort with Highest Survival Rate:", highest_survival_port)
print("Survival Rate:", highest_survival_rate)
```
