```python
# Duck Typing
class Duck:
  def swim(self):
    return "Duck swimming"
  def fly(self):
    return "Duck flying"
```

```python
class Aeroplane:
  def fly(self):
    return "Aeroplane flying"
```

```python
def fly_test(entity):
  return entity.fly()
```

```python
#same function works with different objects
#Python doesn't care what an object's class is. It only cares whether the object has the required method or attribute.
duck = Duck()
fly_test(duck)
```

```python
aeroplane = Aeroplane()
fly_test(aeroplane)
```

```python
# Operator Overloading
class Point:
  def __init__(self,x,y):
    self.x = x
    self.y = y
  def __add__(self,other):
    return Point(self.x + other.x,self.y + other.y)
  def __str__(self):
    return f"({self.x},{self.y})"
```

```python
p1 = Point(2,3)
p2 = Point(4,5)
p3 = p1 + p2
print(p3)
```

```python
#subtraction
class Player:
  def __init__(self, name, health):
    self.name = name
    self.health = health
  def __sub__(self,damage):
    return Player(self.name,self.health - damage)
  def __str__(self):
    return f"{self.name} has {self.health}% health"

p = Player("Smith", 100)
p = p- 70
print(p)
```

```python
# multiplication
class Products:
  def __init__(self, name, price,  quantity):
    self.name = name
    self.price = price
    self.quantity = quantity
  def __mul__(self,other):
    return Products(self.name, self.price * other, self.quantity)
  def __str__(self) -> str:
    return f"{self.name} =  Rs.{self.price}"

p1 = Products("Brush", 100, 2)
total = p1 * p1.quantity
print(total)
```

```python
# division
class Points:
  def __init__(self,x, y):
    self.x = x
    self.y = y

  def __truediv__(self, number):
    return Points(self.x/number, self.y/number)
  def __str__(self):
    return f"{self.x}, {self.y}"

p1 = Points(15,21)
p2 = p1/2
print(p2)
```

```python
#reminder
class product:
  def __init__(self,stock):
    self.stock = stock
  def __mod__(self, box_size):
    return product(self.stock%box_size)
  def __str__(self):
    return f"{self.stock}"
prod = product(1015)
remaining = prod % 10
print(f"Remaining from the stock = {remaining}")
```

```python
#equal to
class BanckAccount:
  def __init__(self, accno, balance):
    self.accno = accno
    self.balance = balance
  def __eq__(self, other):
    return self.accno == other.accno
  def __str__(self):
    return f"{self.accno} = {self.balance}"

a1 = BanckAccount("ACC01",5000)
a2 = BanckAccount("ACC01",5000)
print(a1 == a2)
```

```python
#getitems
class MyList:
  def __init__(self,data):
    self.data = data

  def __getitem__(self, index):
    return self.data[index]

numbers = MyList([10, 20, 30])

print(numbers[0])
print(numbers[2])
```
