```python
import numpy as np

fare_bins = [0,50,100,300,512]
fare_labels = ["Low Fare", "Medium Fare", "High Fare", "Very High Fare"]

titanic_df["FareGroup"] = pd.cut(
    titanic_df["Fare"],
    bins=fare_bins,
    labels=fare_labels,
    include_lowest=True
)
```

```python
#bar chart
#gender count
import matplotlib.pyplot as plt
```

```python
titanic_df["Sex"].value_counts().plot.bar()
plt.title("Gender Distribution")
plt.show()
```

```python
titanic_df["Survived"].value_counts().plot(
    kind =  "bar",
    title = "count of survived passenger",
    figsize= (5,5),
    legend = True,
    yticks = list(range(0,600,50)),
    xlabel = "Survived",
    ylabel = "Count",
    grid = True,
    color = "Green"
)
```

```python
# crosstab
pd.crosstab(titanic_df.Survived, titanic_df.Sex).plot(
    kind = "bar",
    figsize = (5,5),
    title = "Count of survived passenger",
    legend = True,
    xlabel = "Survived (0=No, 1=yes)",
    ylabel = "No. of Passenger",
    grid = True,
    colormap= "Dark2"
)
```

```python
#pie chart
group_by_embarked = titanic_df[["Embarked", "Age"]].groupby("Embarked")
group_by_embarked.count().plot(
    kind = "pie",
    y = "Age",
    figsize = (5,5),
    autopct = "%1.1f%%",
    labels = ["Cherbourg", "Queenstown", "Southampton"],
)
```

```python
pd.crosstab(titanic_df.Survived, titanic_df.Sex).plot(
    kind = "pie",
    fontsize=12,
    figsize = (8,8),
    subplots=True,
    legend = True,
    colormap="Dark2"
)
```

```python
# line graph
titanic_df["Age"].loc[:300].plot.line(figsize=(10,4))
```
