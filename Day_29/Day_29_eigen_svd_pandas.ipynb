```python
import numpy as np
```

```python
# eigen values and eigen vector
a = np.array([
     [1,2],
     [3,4]
])

eigenvalues, eigenvectors = np.linalg.eig(a)

print(eigenvalues)
print(eigenvectors)
```

```python
# Singular value decomposition (SVD)
U, S, V = np.linalg.svd(a)

print("matrix u:")
print(U)
print("matrix s:")
print(S)
print("matrix v:")
print(V)
```
